```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import 'StudentsPerformance.csv'
df = pd.read_csv(f'Assignment12/data/StudentsPerformance.csv')
df.head()
```

```python
df.info()
```

# Task 1

Creating a relational plot with two numerical column and color shading with a categorical column.

```python
sns.relplot(data=df[df['math score']>80], x='math score', y='reading score', hue='gender', kind='line')
plt.show()
```

```python
sns.relplot(data=df, x='math score', y='reading score', hue='test preparation course')
plt.show()
```

# Task 2

Creating a line plot with `sns.lineplot()` and converting the same to scatter style line plot

```python
sns.relplot(data=df, x='math score', y='writing score', col='gender', kind='line')
plt.show()
```

```python
sns.relplot(data=df, x='math score', y='writing score', col='gender')
```

# Task 3

Creating Histogram, KDE, Rugplot for a numerical column

```python
sns.histplot(data=df, x='math score', hue='gender', bins=10, element='step')
```

```python
sns.kdeplot(data=df, x='writing score', hue='gender', fill=True)
```

```python
sns.rugplot(data=df, x='reading score', hue='gender')
plt.show()
```

```python
sns.histplot(data=df, x='reading score', hue='gender', kde=True)
plt.show()
```

# Task 4

Bivariate distribution plots

Creating distribution plots between two numerical values

```python
sns.histplot(data=df, x='math score', y='reading score')
plt.show()
```

```python
sns.kdeplot(data=df, x='math score', y='reading score')
plt.show()
```

# Task 5

Creating a matrix plot

```python
sns.pairplot(data=df, hue='gender')
plt.show()
```

```python
pivoted = df.pivot_table(index='parental level of education', columns='race/ethnicity', values='math score')
sns.heatmap(data=pivoted, annot=True)
plt.show()
```

# Task 6

Categorical plot

```python
sns.barplot(data=df, x='race/ethnicity', y='math score')
plt.show()
```

```python
sns.boxplot(data=df, x='race/ethnicity', y='math score')
plt.show()
```

```python
sns.violinplot(data=df, x='race/ethnicity', y='math score')
plt.show()
```

```python
sns.countplot(data=df, x='race/ethnicity')
plt.show()
```

# Task 7

- Creating a regression plot between two categorical columns
- Creating Implot with hue for a categorical column

```python
sns.regplot(data=df, x='math score', y='reading score')
plt.show()
```

```python
sns.lmplot(data=df, x='math score', y='reading score', hue='gender')
plt.show()
```

# Task 8

- Create a facet plot with two numerical column on x and another on y with one category on left, one on right.
- Create a multiplot dashboard

```python
sns.relplot(data=df, x='reading score', y='writing score', col='gender')
```

```python
sns.jointplot(data=df, x='math score', y='reading score', kind='scatter')
plt.show()
```
